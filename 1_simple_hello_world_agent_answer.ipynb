# Template for Hello World Agent

_Create a minimal (or ‘Hello world’) agent that responds to generic questions both in non-streaming and streaming mode using Microsoft Agent Framework._

```python
# First import class "OllamaChatClient" and "OpenAIChatClient" from "agent_framework.ollama" and 
# "agent_framework.openai", respectively.

# Imports packages

from agent_framework.ollama import OllamaChatClient
from agent_framework.openai import OpenAIChatClient
```

```python
# Set variables over the following information.
# Ollama endpoint is available at http://localhost:11434.
# OpenAI compatible endpoint is available at http://localhost:11434/v1
# Use model Llama 3.2 3B over name "llama3.2:3b"

# Sets endpoints for Ollama models to be available over web requests.

OLLAMA_ENDPOINT="http://localhost:11434"
OLLAMA_ENDPOINT_OPENAI="http://localhost:11434/v1"
OLLAMA_MODEL="llama3.2:3b"
```

## Non-Streaming Mode
_The agents responds in non-streaming mode._

### Using Ollama-Native Chat Client

```python
# Instantiate a native ollama chat client calling constructor of class OllamaChatClient
# passing model name (already stored in a variable)

# Initializes an Ollama chat client
ollama_chat_client = OllamaChatClient(
    model=OLLAMA_MODEL      # The language model that the client serves with
    )

# Creates an agent with Ollama as chat client.
agent = ollama_chat_client.as_agent(
    instructions="You are a helpful general assistant."
)
```

```python
# Pass user query such as "What is the capital of France?" as an argument
# to function "run" of the agent instance.
# Note to call the above function with keyword "await" to pause the execution
# forcing completion of the execution of asynchronous function "run".
# Store the response of the function in a variable and then print the value
# to check the response the model returns against the given query.

result = await agent.run("What is the capital of France?")  # Agent executes the provided message.
print(result)
```

### Using OpenAI Compatible Chat Client

```python
# Now, instead of native chat client, initialiate an OpenAI compatible chat client by passing 
# model name argument to parameter "model", 
# OpenAI compatible endpoint to parameter base_url and
# dummy argument "ollama" to parameter "api_key"
# to the constructor of the class "OpenAIChatClient" and
# store the returned instance of the OpenAI compatible chat client into a variable

# Initializes an OpenAI compatible chat client
openai_chat_client = OpenAIChatClient(
        model="llama3.2:3b",
        base_url="http://localhost:11434/v1",   # URL that the model is available at
        api_key="ollama",                       # Dummy value for OpenAI-compatible local endpoints
    )

# Creates an agent with a chat client compatible to OpenAI (but provided by Ollama)
agent = openai_chat_client.as_agent(
        instructions="You are a helpful general assistant.",
    )
```

```python
# Pass user query such as "What is the capital of France?" as an argument
# to function "run" of the agent instance.
# Note to call the above function with keyword "await".
# Store the response of the function in a variable and then print the value
# to check the response the model returns against the given query.

result = await agent.run("What is the capital of India?")
print(result)
```

## Streaming Mode
_The agents responds in streaming mode._

```python
# Create a asynchronous loop using keyphrase "async for" for the chunks
# that "run" function of agent instance returns.
# Pass argument user query such as "Tell me about Agentic AI." as first parameter,
# argument "True" to parameter "stream" to the function "run" of the agent instance.
# In the body of the loop, check if text property of the chunk object exists or not, 
# and if exists, call function "print" passing
# the text of the chunk object as argument to the first parameter,
# argument blank i.e. "" to the second parameter and 
# value "True" as argument to parameter "flush".

async for chunk in agent.run(
    "Tell me about Agentic AI.", 
    stream=True     # Allows render responses as they get generated, reducing perceived wait times
    ):
    
    if chunk.text:
        print(chunk.text, end="", flush=True)

print()
```
